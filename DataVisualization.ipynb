```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin1")
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar", color="skyblue")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
df.groupby("Region")["Profit"].sum().plot(kind="bar", color="green")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

```python
plt.hist(df["Sales"], bins=30, color="orange")

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Profit"], bins=30, color="purple")

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
sns.heatmap(
    df[["Sales","Profit","Quantity","Discount"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
df["Ship Mode"].value_counts().plot(kind="bar", color="teal")

plt.title("Orders by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Number of Orders")
plt.show()
```

Summary

The visualizations show that Technology has the highest sales, the West region contributes the highest profit, and Standard Class is the most commonly used shipping mode. Sales and profit have a positive relationship, while higher discounts can reduce profitability. These visualizations help understand business performance and support better decision-making.
